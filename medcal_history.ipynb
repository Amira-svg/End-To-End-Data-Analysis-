```python

import numpy as np
import pandas as pd
df=pd.read_csv("D:/Clean data/final_project_cleaned_medical_histo.csv")
pd.set_option('display.max_columns', None)
#pd.set_option('display.max_rows', None)
df["History_ID"]=df["History_ID"].astype(str)
df["Patient_ID"]=df["Patient_ID"].astype(str)
df["Chronic_Disease"]=df["Chronic_Disease"].astype(str)
df["Disease_Severity"]=df["Disease_Severity"].astype(str)
df["Previous_Conditions"]=df["Previous_Conditions"].astype(str)
df=df.drop_duplicates(subset=["History_ID"])
df["Patient_ID"] = pd.to_numeric(df["Patient_ID"], errors="coerce")
case_mask = df["Patient_ID"] > 199999
df.loc[case_mask, "Patient_ID"] = np.random.randint(
    1, 200000, size=case_mask.sum()
)
df.to_csv("D:/بايثون/med_history.csv", index=False)
```
