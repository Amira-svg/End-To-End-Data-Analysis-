```python
import numpy as np
import pandas as pd
df=pd.read_csv("D:/Clean data/final_project raw_patient_treat.csv")
df["Start_Date"]=df["Start_Date"].ffill()
df["End_Date"]=df["End_Date"].bfill()
df=df.drop_duplicates(subset=["Case_ID"])
df["Case_ID"]=df["Case_ID"].astype(str)
df["Treatment_ID"]=df["Treatment_ID"].astype(str)
df["Start_Date"]=pd.to_datetime(df["Start_Date"])
df["End_Date"]=pd.to_datetime(df["End_Date"])
df["Dosage_Group"]=df["Dosage_Group"].astype(str)
df["Treatment_Status"]=df["Treatment_Status"].astype(str)
df=df[df["Start_Date"]<=df["End_Date"]]
df.to_csv("D:/بايثون/patient_treat.csv", index=False)
```
