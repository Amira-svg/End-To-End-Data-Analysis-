```python
import numpy as np
import pandas as pd

df = pd.read_csv(
    "D:/الملف بعد التنظيف الكلي/منكو لله/منكو لله/csv/final_project raw_outcome.csv"
)
pd.set_option("display.max_columns", None)

df["Case_ID"] = pd.to_numeric(df["Case_ID"], errors="coerce")
case_mask = df["Case_ID"] > 600000
df.loc[case_mask, "Case_ID"] = np.random.randint(
    1, 600001, size=case_mask.sum()
)
df["Case_ID"] = df["Case_ID"].fillna(1).astype(int).astype(str)


df["Outcome_ID"] = df["Outcome_ID"].astype(str)
df["Recovery_Date"] = pd.to_datetime(df["Recovery_Date"])


df["Recovery_Days"] = pd.to_numeric(df["Recovery_Days"], errors="coerce")
rec_mask = df["Recovery_Days"] > 39
df.loc[rec_mask, "Recovery_Days"] = np.random.randint(
    1, 40, size=rec_mask.sum()
)
df["Recovery_Days"] = df["Recovery_Days"].fillna(1).astype(int)


df["Hospitalized"] = df["Hospitalized"].astype(bool)
df["ICU_Admission"] = df["ICU_Admission"].astype(bool)
df["Outcome_Status"] = df["Outcome_Status"].astype(str)
df["Side_Effects"] = df["Side_Effects"].astype(str)


df = df.drop_duplicates(subset=["Outcome_ID"])

invalid = (
    (df["Hospitalized"] == True) & (df["Outcome_Status"] != "Hospitalized")
) | ((df["Hospitalized"] == False) & (df["Outcome_Status"] == "Hospitalized"))
df = df[~invalid]

invalid = (df["ICU_Admission"] == True) & (df["Hospitalized"] == False)
df = df[~invalid]

print(df)
df.to_csv("D:/بايثون/Outcome.csv", index=False)
```
